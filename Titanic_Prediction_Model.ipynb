```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.pipeline import Pipeline
```

## Data Importing

```python
titanic_data = pd.read_csv(r"C:\Users\USER\Desktop\webfala\pandas\data\titanic.csv")
titanic_data.head()
```

## Data Preprocessing

```python
titanic_data.info()
```

```python
titanic_data.describe()
```

```python
titanic_data.isnull().sum()
```

```python
titanic_data['Survived'].value_counts()
```

```python
features = ['Pclass','Sex', 'Age', 'SibSp' ,'PaCh','Fare']
target = 'Survived'
```

```python
x = titanic_data[features].copy()
y = titanic_data[target].copy()
```

```python
x
```

```python
x['Sex'] = x['Sex'].map({'female': 1, 'male': 0})
```

```python
x.head()
```

## Data Splitting


```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size =0.2, random_state=42, stratify=y)
```

```python
print('Training data:', x_train.shape)
print('Testing data:', x_test.shape)
```

## Feature Scaling

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

## Logistic Regression Model

```python
logistic_model = LogisticRegression(max_iter=800)

logistic_model.fit(x_train_scaled, y_train)
```

```python
logistic_prediction = logistic_model.predict(x_test_scaled)

logistic_accuracy = accuracy_score(
    y_test,
    logistic_prediction,
)

print("Logistic Regression Accuracy:", logistic_accuracy * 100)
```

```python
print(classification_report(y_test, logistic_prediction))
```

## Decision Tree Model

## Model Training

```python
tree_titanic = DecisionTreeClassifier(
    random_state=42,
    max_depth=3
)

tree_titanic.fit(x_train, y_train)
```

```python
tree_titanic_prediction = tree_titanic.predict(x_test)

tree_titanic_accuracy = accuracy_score(y_test, tree_titanic_prediction)

print('Decision Tree Accuracy', tree_titanic_accuracy * 100)
```

```python
print(classification_report(y_test, tree_titanic_prediction))
```

## Random Forest Model

## Model Training

```python
forest_titanic = RandomForestClassifier(
    max_depth = 5,
    n_estimators = 120,
    random_state = 42
)

forest_titanic.fit(x_train, y_train)
```

```python
forest_titanic_prediction = forest_titanic.predict(x_test)

forest_titanic_accuracy = accuracy_score(y_test, forest_titanic_prediction)

print('Random Forest Accuracy:', forest_titanic_accuracy * 100)
```

```python
print(classification_report(y_test, forest_titanic_prediction))
```

## Support Vector Machine(SVM)

```python
svm_titanic = SVC(
    C= 1000,
    kernel = 'linear',
    random_state=42
    
)

svm_titanic.fit(x_train, y_train)
```

```python
svm_titanic_prediction = svm_titanic.predict(x_test)

svm_titanic_accuracy = accuracy_score(y_test, svm_titanic_prediction)

print('SVM Accuracy:', svm_titanic_accuracy * 100)
```

```python
print(classification_report(y_test, svm_titanic_prediction))
```

## Using Pipeline

```python
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression())
     
])

pipeline.fit(x_train, y_train)
```

```python
y_pred = pipeline.predict(x_test)

pipeline_accuracy = accuracy_score(y_test, y_pred)

print(f"Pipeline Accuracy: {pipeline_accuracy * 100 :.2f}%")
```

```python
print(classification_report(y_test, y_pred))
```

```python
results = pd.DataFrame({
      "Model": [
         'Logistic Regression',
         'Decision Tree',
         'Random Forest',
         'SVM',
         'Pipeline'
      ],
      "Accuracy": [
         logistic_accuracy * 100,
         tree_titanic_accuracy * 100,
         forest_titanic_accuracy * 100,
         svm_titanic_accuracy * 100,
         pipeline_accuracy * 100
      ]
})
results.round(2)
```

## NOTE: Decision Tree has the highest accuracy so i want to improve its performance by tuning

## Decison Tree Tuning

```python
tree_params = {'max_depth': [5, 8, 12]}

tree_grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    tree_params,
    cv=5,
    scoring='accuracy'
)

tree_grid.fit(x_train, y_train)
```

```python
print("Best Parameters:")
print(tree_grid.best_params_)

print("\nBest Cross-Validation Accuracy:")
print(tree_grid.best_score_)
```

```python
best_tree = tree_grid.best_estimator_

tuned_tree_predictions = best_tree.predict(x_test)


tuned_tree_accuracy = accuracy_score(
y_test,
tuned_tree_predictions
)
print("Tuned Decision Tree Accuracy:", tuned_tree_accuracy)
```

```python
print(classification_report(y_test, tuned_tree_predictions))
```
